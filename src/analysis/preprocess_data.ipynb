```python
import json
from collections import Counter

def get_jsonl(path:str):
    with open(path, "r", encoding="utf-8") as f:
        return [json.loads(line) for line in f if line.strip()]

label_0_path = "../../data/processed/label_0.jsonl"
label_1_path = "../../data/processed/label_1.jsonl"

label_0 = get_jsonl(label_0_path)
label_1 = get_jsonl(label_1_path)

print("-----------------------------------")
print("[전체] label 별 데이터 개수")
print("-----------------------------------")
print("label_0:", 
      len(label_0), 
      ">", 
      f"{round(len(label_0)/(len(label_0)+len(label_1))*100, 2)}%")
print("label_1:", 
      len(label_1), 
      ">",
      f"{round(len(label_1)/(len(label_0)+len(label_1))*100, 2)}%")
print()
print("total:", len(label_0)+len(label_1))
print("-----------------------------------")
print("[label 1] detail 별 데이터 개수")
print("-----------------------------------")
detail_counter = Counter(item["detail"] for item in label_1)

for detail, count in sorted(detail_counter.items()):
    print(f"{detail}: {count}")
```

```python
# ! pip install transformers
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("beomi/KcELECTRA-base")

all_data = label_0 + label_1

token_lengths = [
    len(tokenizer(item["full_text"], add_special_tokens=True)["input_ids"])
    for item in all_data
]

print("-----------------------------------")
print("[전체] Token 길이")
print("-----------------------------------")
print(f"max: {max(token_lengths)}")
print(f"min: {min(token_lengths)}")
print(f"avg: {sum(token_lengths)/len(token_lengths):.2f}")

for n in [256, 512, 1024]:
    cnt = sum(l > n for l in token_lengths)
    print(f">{n}: {cnt} ({cnt/len(token_lengths)*100:.2f}%)")

max_idx = token_lengths.index(max(token_lengths))
print("\n가장 긴 샘플")
print(all_data[max_idx]["full_text"])
```
