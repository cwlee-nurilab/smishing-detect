```python
import json
from collections import Counter

def get_jsonl(path:str):
    with open(path, "r", encoding="utf-8") as f:
        return [json.loads(line) for line in f if line.strip()]

label_0_path = "../../data/processed/label_0.jsonl"
label_1_path = "../../data/processed/label_1.jsonl"

label_0 = get_jsonl(label_0_path)
label_1 = get_jsonl(label_1_path)

print("-----------------------------------")
print("[전체] label 별 데이터 개수")
print("-----------------------------------")
print("label_0:", 
      len(label_0), 
      ">", 
      f"{round(len(label_0)/(len(label_0)+len(label_1))*100, 2)}%")
print("label_1:", 
      len(label_1), 
      ">",
      f"{round(len(label_1)/(len(label_0)+len(label_1))*100, 2)}%")
print()
print("total:", len(label_0)+len(label_1))
print("-----------------------------------")
print("[label 1] detail 별 데이터 개수")
print("-----------------------------------")
detail_counter = Counter(item["detail"] for item in label_1)

for detail, count in sorted(detail_counter.items()):
    print(f"{detail}: {count}")
```
